```python
import pandas as pd
from IPython.display import display, HTML

binary_rows = []
multi_rows = []
per_label_tables = []
for name, metrics in results:
    if not metrics:
      continue
    params = metrics.get('params', '')
    sample_n = metrics.get('sample_size', '')
    binary = metrics.get('binary_test') or metrics.get('binary_val')
    if binary:
      row = {'Baseline': name, 'Parameters': params, 'Sample N': sample_n, **binary}
      binary_rows.append(row)
    multi = metrics.get('multi_test') or metrics.get('multi_val')
    if multi:
      base = {'Baseline': name, 'Parameters': params, 'Sample N': sample_n, **{k: v for k, v in multi.items() if k != 'per_label'}}
      multi_rows.append(base)
      per_label = multi.get('per_label', {})
      if per_label:
        df_lbl = pd.DataFrame([
          {
            'Label': lbl,
            'Precision': vals.get('precision', 0.0),
            'Recall': vals.get('recall', 0.0),
            'F1': vals.get('f1', 0.0),
            'Support': vals.get('support', 0)
          }
          for lbl, vals in per_label.items()
        ])
        per_label_tables.append((name, df_lbl))

df_binary = pd.DataFrame(binary_rows)
display(HTML('<strong>Binary classification</strong>'))
display(df_binary)

df_multi = pd.DataFrame(multi_rows)
display(HTML('<strong>Multilabel classification</strong>'))
display(df_multi)

for tbl_name, tbl in per_label_tables:
  display(HTML(f'<strong>{tbl_name} – per-label metrics</strong>'))
  display(tbl)
```

```python
# Extract labels from per_label_tables if available
if per_label_tables:
    labels = per_label_tables[0][1]['Label'].tolist()
else:
    # Fallback: extract from multi_rows if per_label data exists
    labels = []
    for name, metrics in results:
        multi = metrics.get('multi_test') or metrics.get('multi_val')
        if multi and 'per_label' in multi:
            labels = list(multi['per_label'].keys())
            break

# Multi-label performance radar chart
if multi_rows and labels:
    fig, ax = plt.subplots(figsize=(10, 8), subplot_kw=dict(projection='polar'))
    
    angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()
    angles += angles[:1]  # Complete the circle
    
    # Extract per_label data from results
    for name, metrics in results:
        multi = metrics.get('multi_test') or metrics.get('multi_val')
        if multi and 'per_label' in multi:
            per_label = multi['per_label']
            values = [per_label.get(lbl, {}).get('f1', 0) for lbl in labels]
            values += values[:1]  # Complete the circle
            ax.plot(angles, values, 'o-', linewidth=2, label=name, markersize=6)
            ax.fill(angles, values, alpha=0.1)
    
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(labels, fontsize=10)
    ax.set_ylim(0, 1)
    ax.set_title('🎯 Multi-label F1 Performance by Category', pad=20, fontsize=14)
    ax.legend(loc='upper right', bbox_to_anchor=(1.2, 1.0))
    ax.grid(True)
    
    plt.tight_layout()
    plt.show()
else:
    print("No multi-label data available for radar chart")
```
